```python
# importing libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import math as m
```

```python
df = pd.read_csv("dataset/Insurance_data1.csv")

df
```

```python
feature = df[['age']]      # x
target = df[['charges']]  # y
```

```python
# checking relationship

df1 = df.groupby(['age'])[['charges']].mean()

df1
```

```python
# splitting data

x_train,x_test,y_train,y_test = train_test_split(feature,target,test_size=0.2,random_state=42)
```

```python
# model training

linreg = LinearRegression()  # create object

linreg.fit(x_train,y_train)  # model training
```

```python
y_pred = linreg.predict(x_test)
```

```python
# calculate the charges for a person age 50

linreg.predict([[50]])
```

---

# Evalution and Testing

```python
print(y_pred)
```

```python
print(y_test)
```

```python
y_actual = y_test
print(y_actual)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y_actual, y_pred)

print(mae)
```

```python
mse = mean_squared_error(y_actual, y_pred)

print(mse)
```

```python
rmse = m.sqrt(mse)

print(rmse)
```

```python
r2_score = r2_score(y_actual, y_pred)

print(round(r2_score * 100,0))

# model failed
```
