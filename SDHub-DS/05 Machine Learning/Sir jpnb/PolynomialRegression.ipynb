```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("insurance (1).csv")

df
```

## First model using polynomial

```python
x = df[['age','bmi','children']]  # features

y = df['charges']  #target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly1 = PolynomialFeatures(degree=2)
```

```python
x_trainpoly = poly1.fit_transform(x_train)

x_testpoly = poly1.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()

model.fit(x_trainpoly,y_train)
```

```python
y_pred = model.predict(x_testpoly)
```

```python
# model evaluation

from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
r2_score(y_test,y_pred)
```

```python
mse = mean_squared_error(y_test,y_pred)
```

```python
import math as m

rmse = m.sqrt(mse)

rmse
```

## Second model using Poly

```python
df['smoker'] = df['smoker'].map({'yes':1,'no':0})
df['sex'] = df['sex'].map({'male':1,'female':0})

df.head()
```

```python
x = df[['age','sex','bmi','children','smoker']] # features

y = df['charges']   # target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
poly2 = PolynomialFeatures(degree=2)

x_poly = poly2.fit_transform(x_train)
x_testpoly = poly2.transform(x_test)
```

```python
model2 = LinearRegression()

model2.fit(x_poly,y_train)
```

```python
y_pred2 = model2.predict(x_testpoly)
```

```python
r2_score(y_test,y_pred2)
```

```python
mse = mean_squared_error(y_test,y_pred2)

mse
```

```python
rmse = m.sqrt(mse)

rmse
```
